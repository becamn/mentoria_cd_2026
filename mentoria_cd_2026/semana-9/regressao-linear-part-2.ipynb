# Regularização # 

A **regularização** é uma técnica essencial para evitar o **overfitting** (sobreajuste), simplificando o modelo ao impor restrições aos seus pesos.

Na regressão linear, isso é feito adicionando uma penalidade à função de custo, sendo as duas formas mais comuns a **Regularização L2 (Ridge)** e a **Regularização L1 (Lasso).**

Abaixo, há um aprofundamento do funcionamento de cada uma, suas diferenças matemáticas e o impacto na seleção de variáveis, conforme os materiais fornecidos.

### 1. Regressão Ridge (Regularização L2) ###

A regressão Ridge adiciona um termo de penalidade proporcional ao quadrado dos coeficientes à função de erro quadrático médio (MSE).

- **Função de Custo:** 


- **A Norma L2:** O termo de regularização é igual a metade do quadrado da norma ℓ2 do vetor de pesos (denotada por ∥w∥2).

- **O Hiperparâmetro α:** Controla o nível de regularização. Se α=0, temos a regressão linear comum. Se α for muito grande, todos os pesos terminam muito próximos de zero, resultando em uma linha plana que passa pela média dos dados.

- **Impacto (Encolhimento):** A penalidade L2 força o algoritmo de aprendizado não apenas a ajustar os dados, mas também a manter os pesos do modelo o menor possível. Ela encolhe os coeficientes em direção a zero, mas, matematicamente, eles **nunca chegam a ser exatamente zero.**

- **Quando usar:** É a escolha padrão quando você suspeita que a maioria das variáveis explicativas tem alguma influência no resultado ou quando há **multicolinearidade**(variáveis altamente correlacionadas), pois a penalidade L2 reduz a variância dos coeficientes, tornando o modelo mais robusto.


### 2. Regressão Lasso (Regularização L1) ###

O termo Lasso significa *Least Absolute Shrinkage and Selection Operator*. Diferente da Ridge, ela utiliza o valor absoluto dos coeficientes como penalidade.

- **Função de Custo:** 


- **A Norma L1:** Utiliza a **norma ℓ1** do vetor de pesos (também chamada de distância de Manhattan).

- **Impacto (Seleção de Variáveis):** Uma característica crucial da regressão Lasso é que ela tende a **eliminar completamente os pesos das características menos importantes,** definindo-os como zero.

- **Sparsity (Modelos Esparsos):** Devido a essa propriedade, o Lasso realiza automaticamente a **seleção de atributos** *(feature selection)* e produz um modelo esparso, ou seja, um modelo com poucos pesos diferentes de zero.

- **Quando usar:** É preferível quando você suspeita que apenas algumas das variáveis de entrada são realmente úteis para a previsão, permitindo simplificar o modelo final.

### 3. Comparação Visual e Geométrica ###

A diferença fundamental no comportamento de L1 e L2 pode ser entendida pela forma geométrica de suas restrições:

- **L2 (Círculos):** As curvas de nı́vel da penalidade L2 são circulares. O caminho do Gradiente Descendente em direção ao mı́nimo tende a manter os pesos pequenos, mas distribuı́dos entre as variáveis.

- **L1 (Diamantes):** As curvas de nı́vel da penalidade L1 têm formato de diamante (com quinas nos eixos). Durante a otimização, o caminho do gradiente frequentemente atinge um eixo onde um dos coeficientes é exatamente zero e "escorrega" por esse eixo até o mı́nimo global.

### 4. Elastic Net: O Caminho do Meio ###

O **Elastic Net** é um meio-termo que combina as penalidades Ridge e Lasso, permitindo controlar a proporção de cada uma através de um mix ratio *r*.

Nesses casos, o Elastic Net é preferı́vel, pois combina a capacidade de seleção do Lasso com a estabilidade da Ridge.

### 5. Outras Técnicas de Regularização ###

Além das normas L1 e L2, existem outros métodos importantes:

- **Early Stopping:** Utilizado em algoritmos iterativos como o Gradiente Descendente, consiste em interromper o treinamento assim que o erro no conjunto de validação atinge o mínimo, antes que o modelo comece a sofrer overfitting.

- **Dropout:** Comum em Redes Neurais, desativa aleatoriamente uma porcentagem de neurônios durante cada etapa do treinamento, impedindo que os neurônios se tornem excessivamente dependentes uns dos outros.

- **Max-Norm:** Restringe os pesos de entrada de cada neurônio para que a norma L2 seja menor ou igual a um valor máximo determinado.

### 6. Considerações Práticas Essenciais ###

- **Escalonamento de Atributos:** Modelos regularizados são altamente sensíveis à escala dos dados. É obrigatório padronizar os dados (ex: usando *StandardScaler*) antes de aplicar Ridge ou Lasso, para garantir que atributos com escalas maiores não sejam penalizados injustamente.

- **O Hiperparâmetro α (ou λ):** Controla a intensidade da regularização. Se α = 0, temos a regressão linear comum; se α for muito grande, o modelo se torna excessivamente simples (linha plana), aumentando o viés e diminuindo a variância.

- **Regra de Ouro:** A avaliação do modelo deve ser feita sempre com a medida de desempenho **sem regularização** (como o RMSE puro), usando a penalidade apenas durante o treinamento para guiar o aprendizado.

### Resumo de Decisão ###

1. **Evite a Regressão Linear simples:** É quase sempre preferı́vel ter ao menos um pouco de regularização.

2. **Ridge como padrão:** Se você não tem certeza, comece com Ridge.

3. **Lasso/Elastic Net para Seleção:** Se suspeitar que poucas variáveis importam, use Lasso ou Elastic Net para zerar as inúteis.

4. **Elastic Net sobre Lasso:** Use se houver muitas variáveis correlacionadas ou mais variáveis que dados.


**Dica Crucial:** Modelos regularizados são extremamente sensı́veis à escala dos dados. Portanto, é obrigatório padronizar os dados (ex: usando *StandardScaler*) antes de treinar esses modelos, para garantir que atributos com escalas maiores não sejam penalizados injustamente.

### Resumo final ###

Regularização é um mecanismo para evitar **overfitting**, adicionando uma penalidade aos coeficientes do modelo.

- **Ridge (L2):** penaliza o quadrado dos coeficientes → tende a reduzir todos os coeficientes, mas raramente os zera.

- **Lasso (L1):** penaliza o valor absoluto dos coeficientes → pode zerar alguns coeficientes, funcionando como seleção de variáveis

# Entregável avançado # 

Em Python: treinar **3 modelos** no mesmo dataset: Linear simples, Ridge, Lasso; comparar coeficientes (módulo e zeros) e desempenho (R² ou RMSE).

```python
import pandas as pd 
import numpy as np
from sklearn.linear_model import LinearRegression, Ridge, Lasso 
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.model_selection import train_test_split 
```

```python
# Importando a base de dados

df = pd.read_csv("house-price-india.csv")
df
```

```python
# Explorando as colunas

print(df.head())
```

```python
print(df.info())
```

```python
# Supondo que a coluna alvo seja "price" e o resto sejam features

X = df.drop("Price", axis=1)
y = df["Price"]
```

```python
# Divisão treino/teste

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Definir modelos

linear = LinearRegression()
ridge = Ridge(alpha=1.0)
lasso = Lasso(alpha=0.1)
```

```python
# Treino do modelo

linear.fit(X_train, y_train)
ridge.fit(X_train, y_train)
lasso.fit(X_train, y_train)
```

```python
# Previsões do modelo

y_pred_linear = linear.predict(X_test)
y_pred_ridge = ridge.predict(X_test)
y_pred_lasso = lasso.predict(X_test)
```

```python
# Avaliar o desempenho

print("Linear R2:", r2_score(y_test, y_pred_linear))
print("Ridge R2:", r2_score(y_test, y_pred_ridge))
print("Lasso R2:", r2_score(y_test, y_pred_lasso))

print("Linear RMSE:", mean_squared_error(y_test, y_pred_linear, squared=False))
print("Ridge RMSE:", mean_squared_error(y_test, y_pred_ridge, squared=False))
print("Lasso RMSE:", mean_squared_error(y_test, y_pred_lasso, squared=False))
```

**Interpretação:**
Os três modelos tiveram praticamente o mesmo desempenho em termos de variabilidade (R2) e erro médio (RMSE). O que indica que, para esse dataset, a regularização não trouxe ganhos significativos de previsão - o modelo linear simples já estava adequado.


```python
# Comparar coeficientes

coef_df = pd.DataFrame({
    "Linear": linear.coef_,
    "Ridge": ridge.coef_,
    "Lasso": lasso.coef_
}, index=X.columns)

print("\nCoeficientes comparados:")
print(coef_df)
```

**Interpretação:**

No dataset, a comparação apresenta pequenas diferenças de desempenho, mas mudanças claras nos coeficientes.

```python
# Variáveis quantos coeficientes foram zerados pelo Lasso

print("\nCoeficientes zerados no Lasso:", (lasso.coef_ == 0).sum())
```

**Interpretação:**

O Lasso não eliminou nenhuma variável.
