# Naive Bayes e Teorema de Bayes #

O **Teorema de Bayes** é um mecanismo matemático fundamental para **atualizar crenças ou probabilidades** à medida que novas evidências ou dados se tornam disponíveis. Ele permite calcular a probabilidade de um evento acontecer com base no conhecimento prévio de condições que possam estar relacionadas a esse evento.

Onde *P(A|B)* é a probabilidade do evento *A* dado que *B* ocorreu.

### Componentes do Teorema ###

O cálculo bayesiano baseia-se em quatro pilares principais:

- **Probabilidade A Priori (Prior - *P(A)*):** É o nosso conhecimento inicial ou histórico sobre a hipótese antes de coletarmos os dados atuais.

- **Verossimilhança (Likelihood - *P(B∣A)*):** Mede a compatibilidade dos dados observados com a hipótese, respondendo: "Se a minha hipótese estivesse correta, qual seria a probabilidade de observar exatamente estes dados?".

- **Evidência ou Probabilidade Marginal (*P(B)*):** Funciona como um fator de normalização que garante que a soma das probabilidades de todas as hipóteses seja igual a 1 (100%). Frequentemente, esse valor é calculado expandindo o denominador através da **Lei da Probabilidade Total.**

- **Probabilidade A Posteriori (Posterior - *P(A∣B)*):** É o resultado final atualizado, representando a nossa certeza sobre a hipótese após incorporar o conhecimento trazido pelas novas evidências. 


### Importância e Aplicações ###

O teorema é a base da **Inferência Bayesiana,** uma abordagem estatística que trata as probabilidades como subjetivas e as utiliza como um mecanismo formal de atualização de opiniões. Ele demonstra que a probabilidade condicional de um evento A dado B pode ser derivada conhecendo-se a probabilidade de B dado A.

Na Ciência de Dados, este princípio é central para o funcionamento do algoritmo **Naive Bayes.** Esse classificador calcula a probabilidade de uma instância pertencer a cada classe possível e a atribui àquela que apresentar a **Máxima Probabilidade A Posteriori (MAP)**. O Teorema de Bayes também é crucial para evitar erros intuitivos comuns, como o "paradoxo do falso positivo", ao mostrar como taxas basais baixas na população podem influenciar drasticamente a precisão real de testes diagnósticos.

2. **Quais são os principais tipos de Naive Bayes e quando usar cada um?**

**Resposta:**

Existem três variações principais, escolhidas de acordo com a distribuição dos dados de entrada:

- **Gaussian Naive Bayes:** Utilizado para dados numéricos e contínuos. Assume que as variáveis seguem uma distribuição normal ou gaussiana.
    - **Premissa:** Assume que os dados de cada classe seguem uma distribuição normal (gaussiana).
    - **Quando usar:** Em problemas que envolvem medidas físicas ou estatísticas contínuas, como prever classes baseando-se em altura, peso, temperatura ou salários.

- **Multinomial Naive Bayes:** Ideal para frequências ou contagens. É amplamente usado em classificação de texto (como o modelo "Bag of Words") para contar a ocorrência de palavras.

    - **Funcionamento:** A probabilidade é calculada com base na frequência com que determinados eventos (ou termos) ocorrem.
    - **Quando usar:** É o padrão ouro para **classificação de textos,** utilizando o modelo "Bag of Words", onde se conta a quantidade de vezes que cada palavra aparece em um documento para categorizá-lo.

- **Bernoulli Naive Bayes:** Adequado para características binárias (booleanas), onde o interesse reside apenas na presença ou ausência de um atributo.
    - **Funcionamento:** Diferente do Multinomial, ele não se importa com a frequência, mas apenas com a presença ou ausência de um atributo.
    - **Quando usar:** É útil quando as variáveis indicam apenas se algo ocorreu ou não, como na detecção de spam onde se verifica: "a palavra dinheiro aparece no e-mail? Sim (1) ou Não (0)".

3. **Qual é o objetivo do Naive Bayes em classificação?**

**Resposta:**

O objetivo é calcular a probabilidade de uma instância (um conjunto de atributos *X*) pertencer a cada classe possível e, então, atribuir essa instância à classe que apresentar a maior probabilidade. Esse processo é conhecido tecnicamente como critério de **Máxima Probabilidade A Posteriori (MAP).**

4. **O que é probabilidade condicional e por que ela é central no Naive Bayes?**

**Resposta:**

- **O que é:** É a probabilidade de um evento A ocorrer dado que outro evento B já aconteceu, denotada por P(A∣B).
- **Centralidade:** Ela é o coração do algoritmo porque o Naive Bayes classifica os dados estimando justamente a probabilidade condicional da classe dadas as evidências (atributos).


5. **Diferencie prior, likelihood e posterior.**

**Resposta:**

Esses são os três pilares do cálculo bayesiano:

- **Prior (P(A) - Probabilidade A Priori):** É o nosso conhecimento inicial ou histórico sobre a classe antes de analisarmos os novos dados.

- **Likelihood (P(B∣A) - Verossimilhança):** Mede a compatibilidade dos dados observados com a hipótese. Responde: "Se a classe for verdadeira, qual a chance de observar esses atributos?".

- **Posterior (P(A∣B) - Probabilidade A Posteriori):** É a crença atualizada. A probabilidade da classe após incorporar as novas evidências trazidas pelos dados.

**6. Por que o modelo é chamado de “Naive” (Ingênuo)?**

**Resposta:**

Ele recebe esse nome porque assume uma premissa simplista e raramente verdadeira: a de que todos os atributos são completamente independentes entre si, dada a classe. Por exemplo, ao identificar uma fruta, ele assume que o fato de ela ser "vermelha" não tem relação com o fato de ser "redonda", tratando cada evidência isoladamente.

7. **Compare Naive Bayes x SVM.**

**Resposta:**

- **Natureza:** O Naive Bayes é um modelo probabilístico (trabalha com chances e distribuições), enquanto o SVM é um modelo geométrico (busca fronteiras físicas e margens no espaço).

- **Fronteira de Decisão:** O SVM foca nos pontos mais difíceis (vetores de suporte) para maximizar a margem. O Naive Bayes usa todos os dados para construir suas distribuições.

- **Performance:** O Naive Bayes treina de forma extremamente rápida e é excelente para textos e datasets menores. O SVM é mais preciso em dados complexos não lineares, mas escala pior com volumes massivos de dados comparado ao Naive Bayes.


8. **Compare Naive Bayes com Logistic Regression.**

**Resposta:**

- **Gerativo vs. Discriminativo:** O Naive Bayes é um classificador gerativo (tenta modelar como os dados de cada classe são criados). A Regressão Logística é discriminativa (foca em encontrar a linha que melhor separa as classes).

- **Correlação:** Se houver variáveis altamente correlacionadas, o Naive Bayes tende a falhar ou superestimar importâncias devido à sua premissa de independência. A Regressão Logística lida melhor com isso, ajustando os pesos das variáveis de forma conjunta.


9. **Quando você NÃO usaria Naive Bayes?**

**Resposta:**

- **Forte correlação entre atributos:** Quando as variáveis preditoras são muito dependentes umas das outras.Como o modelo recebe o nome de "ingênuo" por assumir a premissa de independência condicional (cada característica é tratada de forma isolada), ele falha ao lidar com variáveis preditoras que são muito dependentes umas das outras
. Em cenários de alta correlação, o Naive Bayes pode superestimar a importância de certas variáveis, prejudicando a precisão da probabilidade posterior. 
- **Interações complexas:** Em problemas onde o resultado é determinado por interações não lineares geométricas complexas entre as variáveis.
- **Necessidade de precisão absoluta:** Em cenários onde outros modelos (como SVM com Kernel ou Boosting) conseguem capturar padrões mais sutis que a "ingenuidade" do Naive Bayes ignora. Ao contrário da Regressão Logística, que ajusta os pesos das variáveis de forma conjunta durante a otimização, o Naive Bayes trata cada evidência separadamente, sendo menos robusto para lidar com redundâncias nos dados.
