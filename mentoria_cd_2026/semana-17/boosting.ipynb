### Revisão ### 

- **Gradiente e Derivadas:** O gradiente indica a direção de maior crescimento de uma função. Na otimização, caminhamos na direção oposta ao gradiente para encontrar o mínimo erro.

- **Funções de Perda:** **MSE** (Erro Quadrático Médio) pune erros grandes severamente. **MAE** (Erro Absoluto Médio) é robusto a outliers. **Log Loss** é a base para classificação probabilística.

- **Underfitting vs Overfitting:** **Underfitting** ocorre quando o modelo é simples demais (alto viés). **Overfitting** ocorre quando ele decora os dados (alta variância).

- **Bias vs Variance:** O objetivo ideal é o equilíbrio *(trade-off)* onde a soma de viés e variância é minimizada, gerando um modelo estável e preciso em dados novos.

### O que é **Boosting** e por que funciona? ###

O **Boosting** (originalmente chamado de hypothesis boosting) refere-se a qualquer método de conjunto capaz de combinar diversos **aprendizes fracos** (weak learners) para criar um **aprendiz forte** (strong learner).

Diferente do Bagging, onde os modelos são treinados de forma independente e paralela, o Boosting funciona através da **aprendizagem sequencial:** cada novo preditor tenta **corrigir o seu antecessor.** Essa abordagem permite que o modelo final reduza tanto o **viés** quanto a **variância**, alcançando uma precisão superior ao focar nos erros cometidos nas etapas anteriores.



### Diferenças entre Bagging e Boosting ###

- **Bagging:** Treina múltiplos modelos em paralelo usando amostras independentes (Bootstrap) e combina os resultados por média ou votação para **reduzir a variância**.

- **Boosting:** Treina modelos de forma sequencial. O objetivo é **reduzir o viés,** fazendo com que cada modelo aprenda com as falhas do anterior. Por ser sequencial, ele é mais difícil de paralelizar do que o Bagging.

### AdaBoost (Adaptive Boosting) ###

O **AdaBoost** foca sua atenção nas instâncias que o predecessor classificou incorretamente. 

- **Reweighting das observações:** No início, todas as instâncias têm o mesmo peso. Após o treino de um modelo, o peso das observações classificadas incorretamente é **aumentado**, forçando o próximo aprendiz a focar nesses "casos difíceis".

- **Influência dos modelos:** Cada preditor recebe um peso (αj) no voto final do conjunto. Quanto mais preciso for o preditor individual, maior será sua influência na decisão final.

- **Weak Learner:** Comumente utiliza o *Decision Stump* (uma árvore de decisão com profundidade 1) como base.

### Gradient Boosting ###

Enquanto o AdaBoost ajusta pesos de instâncias, o **Gradient Boosting** tenta ajustar o novo preditor aos erros **residuais** cometidos pelo predecessor.

- **Residual Learning:** O processo começa com uma predição constante (geralmente a média). A cada passo, uma nova árvore é treinada para prever a diferença entre o valor real e a predição atual (o resíduo).

- **Otimização e Função de Perda:** O algoritmo utiliza o **Gradiente Descendente** para minimizar uma **função de perda** (como o erro quadrático), adicionando modelos gradualmente na direção que reduz o erro total.

### Hiperparâmetros e Overfitting ###

- **Learning Rate (Shrinkage):** Controla a contribuição de cada árvore individual. Valores baixos (ex: 0.1) exigem mais árvores, mas geralmente resultam em melhor capacidade de generalização.

- **Número de Árvores e Profundidade:** Ter árvores demais ou muito profundas pode levar ao **overfitting** (o modelo "decora" o ruído). Por outro lado, poucas árvores podem resultar em **underfitting.**

- **Early Stopping:** Uma técnica essencial para evitar o overfitting; o treinamento é interrompido automaticamente quando o erro no conjunto de validação para de cair e começa a subir.
