# Revisão Regularização # 

### 1. Revisão Regularização e Hiperparâmetros ### 

Lembre-se sempre: 

- **Regularização:** É a te´cnica matemática de penalizar a complexidade do modelo para combater o *overfitting*.

- **Hiperparâmetros:** São as configurações que definimos *antes* do treinamento, que controlam o comportamento do algoritmo (diferente dos parâmetros, que o modelo aprende durante o treino).

Para visualizar por que regularizamos, imagine o "Trade-off" entre viés e variância: A regularização nos ajuda a encontrar o ponto ideal onde o erro de generalização é mínimo.

### 3. Técnicas de Seleção de Hiperparâmetros ### 

Como não sabemos de antemão qual a melhor configuração para um dataset específico, utilizamos técnicas de busca no "espaço de hiperparâmetros".

- **Grid Search (Busca em Grade):** 
    - É uma abordagem exaustiva: você define uma grade de valores possíveis para cada hiperparâmetro e o algoritmo testa todas as combinações possíveis.
    - *Prós:* Garante que você encontrará a melhor combinação dentro da grade definida.
    - *Contras:* Computacionalmente caro se o espaço de busca for muito grande.

- **Randomized Search (Busca Aleatória):**
    - Em vez de testar todas as combinações, ele seleciona aleatoriamente um número definido de combinações para testar.
    - *Prós:* Muito mais rápido e, na prática, frequentemente encontra resultados tão bons ou melhores que o Grid Search, especialmente quando alguns hiperparâmetros são mais influentes que outros.

- **Validação Cruzada (Cross-Validation):**
    - Independentemente da estratégia de busca (Grid ou Random), sempre utilizamos validação cruzada para avaliar as combinações.
    
    - Dividimos os dados de treino em *folds* (partes), treinamos em alguns e validamos em outros, rodando esse processo várias vezes. Isso garante que a nossa escolha de hiperparâmetros não foi um "golpe de sorte" com uma divisão específica dos dados.
