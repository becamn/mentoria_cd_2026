### Revisão: Álgebra Linear ### 

Para a sua revisão de Álgebra Linear e Geometria voltada para Ciência de Dados, organizei os conceitos fundamentais extraídos das fontes:

### Vetores
*   **Definição:** Um vetor é um caso especial de matriz que possui uma única linha (vetor linha) ou uma única coluna (vetor coluna).
*   **Aplicação em ML:** Em aprendizado de máquina, vetores frequentemente representam o **conjunto de características (*features*)** de uma única instância ou cliente, como renda, idade e score de crédito.
*   **Representação:** São geralmente denotados por letras minúsculas em negrito  e representam a posição de uma amostra no espaço de características.

### Matrizes
*   **Definição:** Uma matriz é um arranjo retangular de números, símbolos ou expressões organizados em linhas e colunas.
*   **Uso de Dados:** Um conjunto de dados completo é representado por uma matriz, onde cada linha corresponde a uma instância (um vetor transposto) e cada coluna a uma variável.
*   **Operações:** As operações básicas incluem adição e subtração (entre matrizes de mesmas dimensões), multiplicação por um escalar e a **transposição**, que troca linhas por colunas.
*   **Multiplicação Matricial:** Se A é m x n e B é n x p, o produto AB resulta em uma matriz m x p, sendo essencial para calcular projeções de riscos e previsões de modelos lineares.

### Distância entre Vetores
*   **Distância Euclidiana (Norma L2):** É a medida de distância em linha reta entre dois pontos no espaço n-dimensional, sendo a mais comum para dados contínuos.
*   **Distância Manhattan (Norma L1):** Calcula a soma das diferenças absolutas entre as coordenadas; é frequentemente chamada de "distância de bloco de cidade" e é menos sensível a *outliers* que a Euclidiana.
*   **Distância de Mahalanobis:** Esta medida considera a correlação entre as variáveis e é útil para detectar clientes atípicos em um portfólio.

### Produto Escalar (ou Interno)
*   **Cálculo:** O produto escalar de dois vetores u e v é a soma dos produtos de seus componentes correspondentes.
*   **Significado Geométrico:** Ele mede a projeção de um vetor sobre o outro e está diretamente relacionado ao cosseno do ângulo entre eles.
*   **Importância no SVM:** O produto escalar é usado para determinar de qual lado do hiperplano um ponto de dados está, servindo como base para a classificação.

### Conceito de Redução de Dimensionalidade
*   **Objetivo:** O objetivo é simplificar os dados reduzindo o número de características sem perder informações essenciais.
*   **Benefícios:** Algoritmos de redução tornam o treinamento de outros modelos mais rápido, economizam espaço em disco/memória e permitem a **visualização de dados** complexos em 2D ou 3D para detecção de padrões.
*   **PCA (Análise de Componentes Principais):** É a técnica mais popular; ela identifica eixos (componentes principais) que capturam a maior parte da variabilidade dos dados.
*   **Técnicas Relacionadas:** Além do PCA, existem métodos como LLE (*Locally Linear Embedding*) e t-SNE, este último muito focado na visualização de agrupamentos (*clusters*).
