### Clusterização ###

1. **Conceito de Aprendizado Não Supervisionado**

O aprendizado **não supervisionado** ocorre quando o sistema tenta aprender sem um "professor", utilizando dados de treinamento que **não possuem rótulos** *(unlabeled data).* Ao contrário do supervisionado, onde o modelo busca mapear entradas para saídas conhecidas, aqui o algoritmo tenta descobrir, por conta própria, estruturas, padrões ou agrupamentos ocultos inerentes à própria distribuição dos dados.

2. **O que é Clusterização e Principais Casos de Uso**

A **clusterização** (ou agrupamento) é a tarefa de identificar instâncias semelhantes e atribuí-las a grupos comuns, chamados de *clusters.* O objetivo é que os objetos dentro de um grupo sejam mais parecidos entre si do que com objetos de outros grupos. Principais casos de uso:

- **Segmentação de clientes:** Agrupar clientes por comportamento de compra para campanhas de marketing direcionadas.

- **Análise de dados exploratória:** Descobrir estruturas naturais em novos conjuntos de dados.


- **Redução de dimensionalidade:** Substituir o vetor de características original por um vetor de afinidades com os clusters encontrados.

- **Detecção de anomalias:** Identificar instâncias com baixa afinidade com todos os clusters (como fraudes financeiras).

- **Processamento de imagem:** Segmentar pixels por cor para simplificar imagens ou detectar contornos.

3. **Similaridade vs. Dissimilaridade entre Observações**

A clusterização baseia-se na ideia de que pontos próximos no espaço de características são similares, enquanto pontos distantes são **dissimilares.**

- **Similaridade:** É uma medida de quão próximos dois objetos estão. Em algoritmos como o GMM *(Gaussian Mixture Models)*, utiliza-se o conceito de "afinidade" ou probabilidade de pertencimento.

- **Dissimilaridade:** Geralmente quantificada por métricas de **distância.** Quanto maior a distância entre duas observações, maior a sua dissimilaridade.

4. **Importância da Preparação dos Dados**

A preparação é uma fase crucial (metodologia CRISP-DM) para garantir a qualidade do modelo final. Em clusterização, dados mal preparados podem gerar grupos sem significado prático. Preparar os dados permite:

- Reduzir o impacto de variáveis irrelevantes ou redundantes.

- Garantir que todas as variáveis contribuam de forma equilibrada para o cálculo da similaridade.

- Lidar com valores ausentes *(missings)* que impediriam o cálculo de distâncias matemáticas.


5. **Padronização e Normalização de Variáveis**

Essas técnicas são obrigatórias quando as variáveis possuem escalas diferentes (ex: renda em milhares e idade em dezenas), pois algoritmos baseados em distância são dominados por variáveis de maior magnitude.

- **Normalização (Min-Max Scaling):** Escala os dados para um intervalo fixo, geralmente.

- **Padronização (Z-Score Scaling):** Transforma os dados para terem **média zero e desvio padrão um.** É a técnica preferida para muitos algoritmos, como o PCA e K-Means.

6. **Tratamento de Outliers e seu Impacto**

**Outliers** (valores atípicos) podem distorcer severamente os agrupamentos.

- **Impacto:** Em algoritmos como o K-Means, um único outlier pode "puxar" o centroide para longe da massa principal de dados, tornando os grupos pouco representativos.

- **Tratamento:** As estratégias incluem a remoção das instâncias, transformação dos dados (log ou raiz quadrada) ou uso de algoritmos robustos a outliers, como o **DBSCAN** (que os rotula como ruído) ou o **K-Medoids.**

7. **Distância Euclidiana, Manhattan e Cosseno**

A escolha da métrica define a geometria do cluster:

- **Distância Euclidiana:** É a distância em "linha reta" entre dois pontos. É a mais comum para dados contínuos e padronizados.

- **Distância Manhattan (L1):** Calcula a soma das diferenças absolutas (como caminhar em blocos de uma cidade). É **mais robusta a outliers** que a Euclidiana.

- **Distância do Cosseno:** Mede o cosseno do ângulo entre dois vetores. Ela ignora a magnitude e foca na **orientação**, sendo ideal para dados de alta dimensionalidade ou textos.

8. **Como a escolha da distância influencia os resultados**

A métrica de distância é o que define o "vizinho mais próximo".

- Se você usar a **Euclidiana**, o modelo será muito sensível a escalas e outliers.

- A **Manhattan** pode identificar agrupamentos diferentes em dados com valores extremos.

- A **Distância de Mahalanobis** é superior quando as variáveis são correlacionadas (ex: renda e score de crédito), pois ela utiliza a matriz de covariância para ajustar o espaço.

- O **Cosseno** é essencial se o importante for o perfil do comportamento e não o volume (ex: proporção de gastos em diferentes categorias).
